```python
import torch
from torch import nn
from torch import optim
from model.glove import Glove
from datasets.dataset import Wiki103Dataset
from model import SentimentAnalysisLSTM
from trainer import Trainer
from pipeline import NextWordPredictionPipeline
```

```python
device = 'cuda'
hidden_dim = 512
```

```python
glove = Glove()
model = SentimentAnalysisLSTM(glove.vocab_size, glove.embed_dim,hidden_dim,dtype=torch.float32)
loss_fn = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters())
embedding = nn.Embedding.from_pretrained( 
    glove.weight_matrix,
    freeze=True
).to(torch.float16).to(device)
```

```python
with open("TaskLMTraining/ckpt_epoch_5_batch_600.pth",'rb') as f:
    model.load_state_dict(torch.load(f)["model_state_dict"])
pipeline = NextWordPredictionPipeline(model, glove, embedding)
```

```python
import nltk
from nltk.tokenize import WordPunctTokenizer
import numpy as np
import torch
```

```python


def tokenize(text, tk):
    text = text.lower()
    tokens = tk.tokenize(text)
    return tokens
```

```python
def build_vocab_from_txt(file_path, embedding_dim):
    word2id = {}
    weights = []
    
    # 1. Add Special Tokens (Important for LSTMs)
    # <PAD> = 0 (all zeros), <UNK> = 1 (random numbers)
    word2id['<pad>'] = 0
    weights.append(np.zeros(embedding_dim))
    
    word2id['<unk>'] = 1
    weights.append(np.random.normal(scale=0.6, size=(embedding_dim,)))
    
    # 2. Read the file
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            values = line.split()
            word = values[0]
            vector = np.asarray(values[1:], dtype='float32')
            
            if len(vector) != embedding_dim:
                continue # Skip header if present
                
            word2id[word] = len(weights)
            weights.append(vector)
            
    return word2id, torch.tensor(np.array(weights))
word2id, weight_matrix = build_vocab_from_txt('glove.6B.300d.txt', 300)
```

```python
import json

with open('word2id.json', 'w') as f:
    json.dump(word2id,f)
```

```python
class Glove:
    def __init__(self,pretrained_file = "glove.6B.300d.txt",embed_dim = 300):
        self.word2id, self.weight_matrix = build_vocab_from_txt('glove.6B.300d.txt', embed_dim)
        self.vocab_size = len(self.word2id)
        self.tk = WordPunctTokenizer()
    def decode(self, text:str):
        text = text.lower()
        tokens = self.tk.tokenize(text)
        return [int(self.word2id.get(token,word2id.get('<unk>'))) for token in tokens]
```
